```python
import numpy as np
import pandas as pd
import ast
```

```python
track = pd.read_csv(r'C:\Users\Xsistech\Documents\music recommendation system\track_data_final.csv')
spotify = pd.read_csv(r'C:\Users\Xsistech\Documents\music recommendation system\spotify_data clean.csv')
```

```python
spotify.head()
```

```python
music = pd.concat([spotify, track]) # merging the two datasets together
music.head()
```

```python
music = music[['track_id','album_id','track_name','artist_name','artist_genres','album_name']]
music = music[music['artist_genres'].notna() & (music['artist_genres'] != '[]')]
music['tags'] = music['artist_genres'] + ' ' + music['album_name'] + ' ' + music['artist_name']
music.head()
```

```python
# Removes [, ], and ' from the strings
music['tags'] = music['tags'].str.replace(r"[\[\]',]", "", regex=True)

music['tags'] = music['tags'].apply(lambda x: x.lower())
music['tags']
```

```python
music = music.drop_duplicates('track_name')
music = music.reset_index(drop=True)
music
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
# TfidfVectorizer turns your tags column into a giant grid of numbers. Computers only understand numbers
tfidf = TfidfVectorizer(stop_words='english')
# Ignore the most common English words (the, as, in, a...). Don't count them when you're trying to figure out what this song is about.
tfidf_matrix = tfidf.fit_transform(music['tags']) # transform words into numbers (computer language)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)
# compare every single song against every other song in your dataset
```

```python
def get_recommendations(track_name, cosine_sim=cosine_sim):
  idx = music[music['track_name'] == track_name].index[0]
  sim_scores = list(enumerate(cosine_sim[idx])) # to iterate through a sequence and also keep track of the index of each item
  sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
  sim_scores = sim_scores[1:11] # Get top 10 similar music
  music_indices = [i[0] for i in sim_scores] # extract the music_indices
  return music[['track_name', 'artist_name']].iloc[music_indices]
```

```python
get_recommendations('deja vu')
```

```python
import pickle
with open('music_recommendation_model.pkl', 'wb') as f:
    pickle.dump((music, cosine_sim),f)
```
